```python
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
# Zadacic sa casa 20.10.2026

# Dati su I0 i S, konstantno kroz objekat, diskutovati resenje JPZ za razlicito tau_lambda

I0 = 1.0 

S = 0.5 
```

```python
def formalno_resenje_S_const(I0, S, tau_lambda):
    Iplus = I0 * np.exp(-tau_lambda) + S * (1 - np.exp(-tau_lambda))
    return Iplus
```

```python
log_tau_lambda = np.linspace(-2, 2, 100)
tau_lambda = 10.0 ** log_tau_lambda
```

```python
plt.plot(log_tau_lambda, formalno_resenje_S_const(I0, S, tau_lambda), label='Izlazno zracenje')
plt.hlines(S, log_tau_lambda[0], log_tau_lambda[-1], colors='r', linestyles='dashed', label='Funkcija izvora')
plt.hlines(I0, log_tau_lambda[0], log_tau_lambda[-1], colors='b', linestyles='dashed', label='Ulazno zracenje')
plt.xlabel('log(tau_lambda)')
plt.ylabel('Iplus')
plt.title('Formalno resenje za konstantno S')
plt.legend()
```
